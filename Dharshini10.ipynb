```python
data.fillna({
    col:data[col].mode()[0]
    for col in data.columns
    if data[col].isnull().any()
},inplace=True)
print(data.isnull().sum())
```

```python

data.drop_duplicates(inplace=True)

if "Date" in data.columns:
    data["Date"] = pd.to_datetime(data["Date"], errors="coerce")

print(data.dtypes)
print(data.head())
```

```python

print(data.loc[0:4, :])


print(data.loc[:, ["Crop", "Production"]])


print(data.iloc[0:5, :])

print(data.iloc[0:10, 0:5])
```

```python

print(data[data["Production"] > 1000])


print(data[data["Production"] < 1000])

print(data[data["Production"] >= 1000])


print(data[(data["Production"] > 1000) &
         (data["Crop"] == "Rice")])


print(data[(data["Production"] > 1000) |
         (data["Crop"] == "Rice")])

print(data[~(data["Crop"] == "Rice")])
```

```python

data["Production_in_Thousands"] = data["Production"] / 1000

print(data.head())
```

```python

print(data.groupby("Crop")["Production"].sum())


print(data.groupby("Crop")["Production"].mean())


print(data.groupby("Crop")["Production"].agg(
    ["min", "max", "sum", "mean", "count", "std"]
))
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

data["Crop_Encoded"] = le.fit_transform(data["Crop"])

print(data[["Crop", "Crop_Encoded"]].head())
```

```python
data_onehot = pd.get_dummies(
    data,
    columns=["Season"],
    dtype=int
)

print(data_onehot.head())
```

```python

from sklearn.preprocessing import OrdinalEncoder


print(data.columns)


order = [["Low", "Medium", "High"]]


oe = OrdinalEncoder(
    categories=order,
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

data["Production_Level_Encoded"] = oe.fit_transform(
    data[["Production_Level"]].astype(str)
)


print(data[["Production_Level", "Production_Level_Encoded"]].head())
```

```python
from sklearn.preprocessing import OrdinalEncoder


data["Season"] = data["Season"].str.strip()


order = [[
    "Autumn",
    "Kharif",
    "Rabi",
    "Summer",
    "Winter",
    "Whole Year"
]]


oe = OrdinalEncoder(categories=order)


data["Season_Encoded"] = oe.fit_transform(
    data[["Season"]]
)


print(data[["Season", "Season_Encoded"]].head())
```

```python
data = pd.to_numeric(data["Production"], errors="coerce").dropna()

print(data.head())
```

```python
print("Mean:", data.mean())
print("Median:", data.median())
print("Mode:", data.mode().tolist())
```

```python
print("Standard Deviation:", data.std())
print("Variance:", data.var())
```

```python
minimum = data.min()
maximum = data.max()
range_value = maximum - minimum

print("Minimum:", minimum)
print("Maximum:", maximum)
print("Range:", range_value)
```

```python
Q1 = data.quantile(0.25)
Q2 = data.quantile(0.50)
Q3 = data.quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
print("Skewness:", data.skew())
```

```python
print("Kurtosis:", data.kurt())
```

```python
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = data[
    (data < lower_limit) |
    (data > upper_limit)
]

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Outliers:")
print(outliers)
print("Number of Outliers:", len(outliers))
```

```python
mean_value = data.mean()
std_value = data.std()

if mean_value != 0:
    CV = (std_value / abs(mean_value)) * 100
    print("Coefficient of Variation:", CV, "%")
else:
    print("CV cannot be calculated because mean is zero.")
```

```python
mean_value = data.mean()
median_value = data.median()

print("Mean:", mean_value)
print("Median:", median_value)

if mean_value > median_value:
    print("Distribution is likely positively skewed.")
elif mean_value < median_value:
    print("Distribution is likely negatively skewed.")
else:
    print("Mean and median are equal; distribution may be symmetric.")
```
