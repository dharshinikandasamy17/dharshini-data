```python
tips = sns.load_dataset('tips')
numeric_cols = tips.select_dtypes(include='number').columns
```

```python
scaler = MinMaxScaler()
tips_normalized = pd.DataFrame(scaler.fit_transform(tips[numeric_cols]),columns=numeric_cols)
print(tips_normalized.head())
```

```python
num_cols = ['total_bill', 'tip', 'size']
cat_cols = ['sex', 'smoker', 'day', 'time']
for col in num_cols:
    tips[col] = tips[col].fillna(tips[col].median())
for col in cat_cols:
    tips[col] = tips[col].fillna(tips[col].mode()[0])
print(tips.isnull().sum())
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

tips['sex'] = le.fit_transform(tips['sex'])
tips['smoker'] = le.fit_transform(tips['smoker'])
tips['day'] = le.fit_transform(tips['day'])
tips['time'] = le.fit_transform(tips['time'])

print(tips.head())
```

```python
print("Sex Frequency:")
print(tips['sex'].value_counts())

print("Smoker Frequency:")
print(tips['smoker'].value_counts())

print("Day Frequency:")
print(tips['day'].value_counts())

print("Time Frequency:")
print(tips['time'].value_counts())
```

```python
print("Sex Percentage:")
print(tips['sex'].value_counts(normalize=True) * 100)

print("Smoker Percentage:")
print(tips['smoker'].value_counts(normalize=True) * 100)

print("Day Percentage:")
print(tips['day'].value_counts(normalize=True) * 100)

print("Time Percentage:")
print(tips['time'].value_counts(normalize=True) * 100)
```

```python
print("Most Common Sex:", tips['sex'].mode()[0])

print("Most Common Smoker:", tips['smoker'].mode()[0])

print("Most Common Day:", tips['day'].mode()[0])

print("Most Common Time:", ['time'].mode()[0])
```
